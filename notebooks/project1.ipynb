```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/raw/Sample - Superstore.csv", encoding="latin-1")
```

```python
print(df.head())
print(df.info())
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
df_clean = df.dropna()
```

```python
df_clean.to_csv("../data/processed/data_clean.csv", index=False)
```

```python
total_sales = df_clean["Sales"].sum()
print("Total sales:", total_sales)
```

```python
avg_profit = df_clean["Profit"].mean()
print("Average profit:", avg_profit)
```

```python
plt.figure(figsize=(10,5))

df_clean.groupby("Category")["Sales"].sum().plot(kind="bar")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.hist(df_clean["Profit"], bins=100, edgecolor='black')  

plt.title("Profit Distribution")
plt.xlabel("Profit")
plt.ylabel("Frequency")

plt.yscale('log')
plt.xlim(-2000, 2500)

plt.grid(True)
plt.show()
```
